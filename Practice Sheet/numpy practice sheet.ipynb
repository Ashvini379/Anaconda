## 6. Broadcasting

Broadcasting lets NumPy combine arrays of **different shapes** without copying data.

The rules, applied to the shapes right-aligned:

1. If the arrays have different `ndim`, the smaller shape is padded with `1`s on the **left**.
2. Two dimensions are compatible if they are **equal** or one of them is **1**.
3. A dimension of size 1 is stretched to match the other.

```
(3, 4)  +  (4,)      ->  (3, 4)   OK   [(4,) becomes (1,4) then stretches]
(3, 1)  +  (1, 4)    ->  (3, 4)   OK
(3, 4)  +  (3,)      ->  ERROR         [4 vs 3 - neither is 1]
```
