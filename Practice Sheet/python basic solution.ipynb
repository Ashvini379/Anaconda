**Q14 — Palindrome checker.**

Write `is_palindrome(text)` that returns `True` if `text` reads the same forwards and backwards,
**ignoring case and spaces**.

`"Never odd or even"` &rarr; `True`

````python
def is_palindrome(text):
    cleaned = text.replace(" ", "").lower()
    return cleaned == cleaned[::-1]


assert is_palindrome("racecar") is True
assert is_palindrome("Never odd or even") is True
assert is_palindrome("Madam") is True
assert is_palindrome("python") is False
print("Q14 passed")
````

**Q15 — Count the vowels.**

Write `count_vowels(text)` returning how many vowels (`a e i o u`, either case) the text contains.

````python
def count_vowels(text):
    count = 0
    for ch in text.lower():
        if ch in "aeiou":
            count += 1
    return count

# One-liner alternative:
# return sum(1 for ch in text.lower() if ch in "aeiou")


assert count_vowels("hello") == 2
assert count_vowels("PYTHON") == 1
assert count_vowels("AEIOU aeiou") == 10
assert count_vowels("rhythm") == 0
print("Q15 passed")
````

**Q17 — Password validator.**

Write `validate_password(pwd)` that returns `True` only when **all** of these hold:

- length is at least 8
- contains at least one uppercase letter
- contains at least one lowercase letter
- contains at least one digit

Hint: `ch.isupper()`, `ch.islower()`, `ch.isdigit()`.

````python
def validate_password(pwd):
    if len(pwd) < 8:
        return False
    has_upper = any(ch.isupper() for ch in pwd)
    has_lower = any(ch.islower() for ch in pwd)
    has_digit = any(ch.isdigit() for ch in pwd)
    return has_upper and has_lower and has_digit


assert validate_password("Passw0rd") is True
assert validate_password("Pass0rd") is False      # too short
assert validate_password("password1") is False    # no uppercase
assert validate_password("PASSWORD1") is False    # no lowercase
assert validate_password("PasswordX") is False    # no digit
print("Q17 passed")
````

**Q18 — Longest word.**

Write `longest_word(sentence)` that returns the longest word. If two words tie, return the one that
appears **first**.

````python
def longest_word(sentence):
    longest = ""
    for word in sentence.split():
        if len(word) > len(longest):
            longest = word
    return longest

# Alternative: return max(sentence.split(), key=len)


assert longest_word("I love programming in Python") == "programming"
assert longest_word("one two six") == "one"
assert longest_word("short longerword mid") == "longerword"
print("Q18 passed")
````

## 3. Lists & Tuples

**Lists** are ordered and **mutable** &mdash; `[]`.

Use a tuple when the collection is a fixed record (a coordinate, a database row); use a list when the

**Q23 — Second largest.**

Write `second_largest(nums)` returning the second largest **distinct** value.
Return `None` if there is no such value.

`[4, 9, 9, 2]` &rarr; `4`   (because 9 is the largest, duplicates don't count twice)

````python
def second_largest(nums):
    unique = sorted(set(nums), reverse=True)
    if len(unique) < 2:
        return None
    return unique[1]


assert second_largest([10, 4, 78, 33, 78]) == 33
assert second_largest([4, 9, 9, 2]) == 4
assert second_largest([5, 5, 5]) is None
assert second_largest([7]) is None
print("Q23 passed")
````

**Q24 — Remove duplicates, keep the order.**

Write `remove_duplicates(items)` that returns a new list with duplicates dropped, keeping the
**first** appearance of each value.

(`set(items)` removes duplicates but loses the order &mdash; so you need a loop.)

````python
def remove_duplicates(items):
    result = []
    for item in items:
        if item not in result:
            result.append(item)
    return result


assert remove_duplicates([1, 2, 2, 3, 1, 4]) == [1, 2, 3, 4]
assert remove_duplicates(["a", "b", "a", "c"]) == ["a", "b", "c"]
assert remove_duplicates([]) == []
print("Q24 passed")
````

**Q30 — Grade calculator (if / elif / else).**

Write `grade(marks)`:

| Marks | Grade |
|---|---|
| 90–100 | `"A"` |
| 80–89 | `"B"` |
| 70–79 | `"C"` |
| 60–69 | `"D"` |
| below 60 | `"F"` |
| outside 0–100 | `"Invalid"` |

````python
def grade(marks):
    if marks < 0 or marks > 100:
        return "Invalid"
    elif marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"


assert grade(95) == "A" and grade(90) == "A"
assert grade(85) == "B" and grade(72) == "C" and grade(60) == "D"
assert grade(59) == "F" and grade(0) == "F"
assert grade(101) == "Invalid" and grade(-5) == "Invalid"
print("Q30 passed")
````

**Q33 — while loop: sum of digits.**

Write `digit_sum(n)` using a **while** loop (no `str()` conversion) that adds up the digits of a
positive integer.

`1234` &rarr; `10`

````python
def digit_sum(n):
    total = 0
    while n > 0:
        total += n % 10    # last digit
        n //= 10           # chop the last digit off
    return total


assert digit_sum(1234) == 10
assert digit_sum(9) == 9
assert digit_sum(1000) == 1
assert digit_sum(0) == 0
print("Q33 passed")
````

Write `reverse_number(n)` that reverses the digits of a positive integer, **without** converting to a
string.

`1234` &rarr; `4321`, `1200` &rarr; `21`

````python
def reverse_number(n):
    reversed_n = 0
    while n > 0:
        reversed_n = reversed_n * 10 + n % 10
        n //= 10
    return reversed_n


assert reverse_number(1234) == 4321
assert reverse_number(7) == 7
assert reverse_number(1200) == 21
print("Q34 passed")
````

**Q36 — break and continue.**

Two small functions:

1. `is_prime(n)` &rarr; `True` if `n` is prime (`n < 2` is not prime).
2. `first_prime_after(n)` &rarr; the smallest prime **strictly greater** than `n`.
   Use a `while True:` loop with a `break`.

````python
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False       # early exit - no need to test further
    return True

def first_prime_after(n):
    candidate = n + 1
    while True:
        if is_prime(candidate):
            break              # found it, leave the loop
        candidate += 1
    return candidate


assert is_prime(2) is True and is_prime(13) is True
assert is_prime(1) is False and is_prime(9) is False
assert first_prime_after(10) == 11
assert first_prime_after(14) == 17
assert first_prime_after(1) == 2
print("Q36 passed")
````

**Q37 — Nested loops: star triangle.**

Write `triangle(n)` returning a list of `n` strings, where row `i` contains `i` stars:

```
*

***
****
```
&rarr; `["*", "**", "***", "****"]`

Write it with a nested loop first (outer row, inner column), then try the `"*" * i` shortcut.

````python
def triangle(n):
    rows = []
    for i in range(1, n + 1):
        row = ""
        for _ in range(i):      # inner loop builds one row
            row += "*"
        rows.append(row)
    return rows

# Shortcut: return ["*" * i for i in range(1, n + 1)]


assert triangle(4) == ["*", "**", "***", "****"]
assert triangle(1) == ["*"]
assert triangle(0) == []
print("Q37 passed")
````

**Q38 — for ... else.**

Write `find_index(items, target)` that returns the index of `target`, or `-1` if it is absent.

Use the `for ... else` construct: the `else` block runs **only if the loop finished without
hitting `break`**.

````python
def find_index(items, target):
    for i in range(len(items)):
        if items[i] == target:
            return i
    else:
        return -1       # reached only when the loop was never broken/returned out of


assert find_index([4, 8, 15, 16], 15) == 2
assert find_index([4, 8, 15, 16], 99) == -1
assert find_index([], 1) == -1
assert find_index(["a", "b"], "a") == 0
print("Q38 passed")
````

**Q42 — *args and **kwargs.**

1. `total(*args)` &rarr; the sum of any number of numbers (`0` when called with none).

   in the order they were passed.

`describe(name="Asha", age=21)` &rarr; `"name=Asha, age=21"`

````python
def total(*args):
    # args is a TUPLE of everything passed positionally
    return sum(args)

def describe(**kwargs):
    # kwargs is a DICT of everything passed by keyword
    return ", ".join(f"{key}={value}" for key, value in kwargs.items())


assert total(1, 2, 3) == 6
assert total() == 0
assert total(*[10, 20, 30]) == 60
assert describe(name="Asha", age=21) == "name=Asha, age=21"
assert describe() == ""
print("Q42 passed")
````

**Q44 — Recursion.**

Write two recursive functions (no loops):

- `fact(n)` &rarr; `n!`, with `fact(0) == 1`
- `fib(n)` &rarr; the n-th Fibonacci number, where `fib(0) == 0` and `fib(1) == 1`

Every recursive function needs a **base case** or it will never stop.

````python
def fact(n):
    if n <= 1:          # base case
        return 1
    return n * fact(n - 1)

def fib(n):
    if n < 2:           # base cases: fib(0)=0, fib(1)=1
        return n
    return fib(n - 1) + fib(n - 2)


assert fact(0) == 1 and fact(1) == 1
assert fact(5) == 120 and fact(7) == 5040
assert [fib(i) for i in range(8)] == [0, 1, 1, 2, 3, 5, 8, 13]
print("Q44 passed")
````

- `bump_global()` should use the `global` keyword to increase the real `counter` by 1 and return it.

````python
counter = 0

def bump_local():
    counter = 99        # creates a NEW local name; the global is untouched
    return counter

def bump_global():
    global counter      # says "use the module-level counter"
    counter += 1
    return counter


assert bump_local() == 99
assert counter == 0, "the global should not have changed"
assert bump_global() == 1
assert bump_global() == 2
assert counter == 2
print("Q46 passed")
````

**Q50 — Text analyser.**

Write `analyse(text)` returning a tuple `(word_count, char_count, longest, most_common)` where:

- `word_count` &rarr; number of words
- `char_count` &rarr; number of characters **excluding spaces**
- `longest` &rarr; the longest word (lowercased, punctuation-free)
- `most_common` &rarr; the most frequently occurring word (lowercased)

Strip `.,!?` from words before counting.

````python
def analyse(text):
    raw_words = text.split()
    words = [w.strip(".,!?").lower() for w in raw_words]

    word_count = len(words)
    char_count = len(text.replace(" ", ""))
    longest = max(words, key=len)
    most_common = max(words, key=words.count)   # ties -> first one found

    return word_count, char_count, longest, most_common


res = analyse("the quick brown fox jumps over the lazy dog the end")
assert res[0] == 11
assert res[2] == "quick"
assert res[3] == "the"
assert analyse("Hello world!")[1] == len("Helloworld!")
print("Q50 passed")
````

**Q51 — Shopping cart.**

`cart` is a list of `(item, unit_price, quantity)` tuples.

Write `checkout(cart, discount=0)` returning a tuple `(subtotal, total, priciest_line)`:

- `subtotal` &rarr; sum of `unit_price * quantity` across all lines, rounded to 2 decimals
- `total` &rarr; subtotal after applying `discount` as a **percentage**, rounded to 2 decimals
- `priciest_line` &rarr; the **item name** of the line with the highest line value

````python
cart = [("notebook", 45.50, 3),
        ("pen", 10.00, 10),
        ("backpack", 899.00, 1)]

def checkout(cart, discount=0):
    subtotal = round(sum(price * qty for _, price, qty in cart), 2)
    total = round(subtotal * (1 - discount / 100), 2)
    priciest_line = max(cart, key=lambda line: line[1] * line[2])[0]
    return subtotal, total, priciest_line

print(checkout(cart))
print(checkout(cart, discount=10))


sub, tot, top = checkout(cart)
assert sub == 1135.5
assert tot == 1135.5
assert top == "backpack"
assert checkout(cart, 10)[1] == 1021.95
print("Q51 passed")
````

**Q52 — Number classifier.**

Write `classify(nums)` returning a tuple `(positives, negatives, zeros, evens, odds)` &mdash;
five **counts**. Treat `0` as even.

````python
def classify(nums):
    positives = negatives = zeros = evens = odds = 0
    for n in nums:
        if n > 0:
            positives += 1
        elif n < 0:
            negatives += 1
        else:
            zeros += 1

        if n % 2 == 0:
            evens += 1
        else:
            odds += 1
    return positives, negatives, zeros, evens, odds


assert classify([1, -2, 0, 4, -5, 7]) == (3, 2, 1, 3, 3)
assert classify([]) == (0, 0, 0, 0, 0)
assert classify([0]) == (0, 0, 1, 1, 0)
print("Q52 passed")
````
