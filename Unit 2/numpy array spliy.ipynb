## Array Concatenation and Splitting

All of the preceding routines worked on single arrays. It's also possible to combine multiple arrays into one, and to conversely split a single array into multiple arrays. We'll take a look at those operations here.

### Concatenation of arrays

Concatenation, or joining of two arrays in NumPy, is primarily accomplished using the routines ``np.concatenate``, ``np.vstack``, and ``np.hstack``.
``np.concatenate`` takes a tuple or list of arrays as its first argument, as we can see here:

``axis=0 → across rows, down the vertical direction.
axis=1 → across columns, horizontal direction.``

````python
x = np.array([1, 2, 3])
y = np.array((3, 2, 1))
print(x.shape, y.shape, np.concatenate([x, y],  axis = 0))
````

``Because x and y are both 1D arrays of shape (3,). Concatenation happens along that single axis.``

#### If i try to reshape it and make concatenation, it creates Axiserror

````python
x.reshape(1,3)
y.reshape(1,3)
print(x.shape, y.shape)
np.concatenate([x, y], axis = 1)
````

```
x.reshape(1,3) → creates a reshaped view of x, but you didn’t assign it back.
x is still shape (3,), not (1,3).
Similarly, y.reshape(1,3) → but y remains shape (3,).
```

````python
x = x.reshape(1,3)
y = y.reshape(1,3)
np.concatenate([x, y])
````

```
concatenate → works on existing axes only.
stack / vstack → create a new axis (row-wise stacking).
```

````python
x.shape
````

````python
x2 = np.arange(1, 10).reshape((3, 3))
x2
````

````python
np.concatenate([x, x2], axis = 0)
````

You can also concatenate more than two arrays at once:

````python
z = np.array([99, 99, 99]).reshape(1,3)
print(np.concatenate([x,y,x2,z]))
````

It can also be used for two-dimensional arrays:

````python
grid = np.array([[1, 2, 3],
                 [4, 5, 6]])
````

````python
# concatenate along the first axis
np.concatenate([grid, grid])
````

````python
# concatenate along the second axis (zero-indexed)
np.concatenate([grid, grid], axis=1)
````

For working with arrays of mixed dimensions, it can be clearer to use the ``np.vstack`` (vertical stack) and ``np.hstack`` (horizontal stack) functions:

````python
x = np.array([1, 2, 3])
grid = np.array([[9, 8, 7],
                 [6, 5, 4]])

# vertically stack the arrays

print(x.shape, grid.shape)
np.vstack([x, grid])
````

````python
# horizontally stack the arrays
y = np.array([[99],
              [99]])

print(y.shape, grid.shape)
np.hstack([grid, y])
````

### Splitting of arrays

The opposite of concatenation is splitting, which is implemented by the functions ``np.split``, ``np.hsplit``, and ``np.vsplit``.  For each of these, we can pass a list of indices giving the split points:

``numpy.split(ary, indices_or_sections, axis=0)``

````python
x = [1, 2, 3, 99, 99, 3, 2, 1, 5]
x1, x2, x3 = np.split(x, [3,6], axis =0)
print(x1, x2, x3)
````

````python
x1, x2, x3 = np.split(grid, [1, 2], axis =1)
print(x1)
print(x2)
print(x3)
````

Notice that *N* split-points, leads to *N + 1* subarrays.
The related functions ``np.hsplit`` and ``np.vsplit`` are similar:

````python
grid = np.arange(16).reshape((4, 4))
grid
````

````python
upper, lower = np.vsplit(grid, [2])
print("Upper = " , upper)
print("Lower = " , lower)
````

````python
left, right = np.hsplit(grid, [2])
print(left)
print(right)
````

# Write an Array to a CSV File in Python Using the numpy.savetxt() Method

The ``numpy.savetxt(fname, array, delimiter=)`` method saves the input array to the file ``- fname``. The ``delimiter`` argument is the value used as the separator to separate the independent value in the data.

To save the array to a CSV file, we will need to pass the ``fname`` with the ``.csv`` extension and use the comma ``,`` as the separator. The below example code demonstrates how to use the ``numpy.savetxt()`` to write an array to a CSV file in Python.``numpy.loadtxt()`` can be used to laod data into notebook.

````python
# a = np.array([[1,4,2],[7,9,4],[0,6,2]])
np.savetxt('Array1.csv', grid, delimiter=',', fmt="%d")
````

````python
arr = np.loadtxt("Array1.csv", delimiter=",", dtype = int)
print(arr)  #all numbers are read as floating-point values, 
                # unless you explicitly tell NumPy.
````

````python
arr = np.loadtxt("Array1.csv", delimiter=",").astype(np.int64)
print(arr)
````

# Write Array to a CSV File in Python Using the writer.writerows() Method 

````python
import csv
import numpy

arr = numpy.array([[7,9,1],[5,6,3],[1,4,3]])

with open('Array1.csv', 'a') as file:
    mywriter = csv.writer(file, delimiter=',')
    mywriter.writerows(arr)
````

To read the array from the CSV file, we can use the ``csv.reader(file)`` method’s ``writer`` object to read the array from the ``file`` object file of the CSV file in Python.

````python
import csv

with open('Array1.csv', 'r', newline='') as file:
      myreader = csv.reader(file, delimiter=',')
      for rows in myreader:
        print(rows)
````

``` 
csv.reader(f) → reads rows as lists of strings (["10", "20", "30"]).
map(int, row) → converts each element of the row to an integer.
```

````python
import csv

with open('Array1.csv', 'r', newline='') as file:
    reader = csv.reader(file)
    
    # Convert each row (list of strings) → list of ints
    data = [list(map(int, row)) for row in reader]

# print(data)

    for i in data:
        print(i)
````
