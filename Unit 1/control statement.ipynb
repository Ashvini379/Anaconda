<h3>break and continue</h3>

```python
x = 0

while x < 10:
    print("x is currently: ", x)
    x+=1
    if x == 3:
        print("Breaking because x==3")
        break
    else:
        print("Continuing....")
        continue
```

Task: Use the division and modulus method to get the sum of digits of a 5-digit number, without converting it to a string:

```python
temp = "12345"
n = 0

for i in temp:
    n += int(i)
    
print(n)   
```

```python
temp = 12345
total = 0

while temp > 0:
        digit = temp % 10   # Get last digit
        total += digit      # Add digit to sum
        temp //= 10         # Remove last digit
    
print("Sum of digits:", total)
```

<h3>Enumerate</h3>-- without having to manually create and update a counter.

```python
index_count = 0

for letter in 'abcde':
    print("At index {} the letter is {}". format(index_count,letter))
    index_count += 1
```

```python
for i,letter in enumerate('abcde',3):
    print("At index {} the letter is {}". format(i,letter))
```

```python
list(enumerate('abcde',5))
```

<h3> random </h3>

Python comes with a built-in random library. There are a lot of functions included in the random library. Here we will look into few important functions which will be widely used in data science.

```python
from random import shuffle
```

```python
mylist = list(range(0,11))
mylist
```

```python
tot = sum(mylist)
print(tot)
```

```python
shuffle(mylist)
mylist
```

```python
from random import randint
```

```python
# Returns random integer in the range [a, b], including both end points
randint(0, 100)
```

```python
from random import randrange
print(randrange(0, 10))
```

```python
mylist = []

for i in range(0,100):
    x = randint(1,100)
    mylist.append(x)

print(mylist)
    
```

```python
from random import sample
x = sample(range(1,101), 100)
```

```python
print(x)
```

```python
# To choose random numbers from the list
x = sample(mylist,3)
print(x)
```
