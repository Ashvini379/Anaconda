# Connecting to a Database

If you have installed Python from the official Python.org installer (or Anaconda), then yes — Python comes with the built-in sqlite3 module.

````python
import sqlite3

# Connect to database (creates if it doesn’t exist)
conn = sqlite3.connect("example.db")

# Create a cursor object (used to execute SQL commands)
cursor = conn.cursor()
````

## Creating a Table

````python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
     id INTEGER PRIMARY KEY AUTOINCREMENT,
     name TEXT NOT NULL,
     age INTEGER,
     grade TEXT
)
""")
````

## Inserting Data

````python
cursor.execute("INSERT INTO students (name, age, grade) VALUES (?, ?, ?)", 
               ("Utkrishta", 20, "A"))  # ?--> placeholder for parameters.
cursor.execute("INSERT INTO students (name, age, grade) VALUES (?, ?, ?)", 
               ("Greeshma", 22, "B"))

conn.commit()   # Save changes
````

```
First ? → "Utkrishta"
Second ? → 20
Third ? → "A"
```

## Fetching Data

````python
cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()

for i in rows:
    print(i)
````

## Updating Data

````python
cursor.execute("UPDATE students SET grade = ? WHERE name = ?", ("A+", "Greeshma"))
conn.commit()
````

## Deleting Data

````python
cursor.execute("DELETE FROM students WHERE name = ?", ("Utkrishta",))
conn.commit()
````

## Closing the Connection

````python
conn.close()
````

# Loading Data from Database

````python
import sqlite3

# Connect to the database (or create one if it doesn't exist)
conn = sqlite3.connect('sample.db') 
````

````python
# Create table if it doesn't exist
cursor = conn.cursor()
cursor.execute('''
    CREATE TABLE IF NOT EXISTS users (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        age INTEGER
    )
''')
conn.commit()
````

````python
# List of tuples to insert
data = [
    ('Aanshi', 5),
    ('Sreeja', 35),
    ('Raghav', 10)
]

# Use ? for SQLite parameter substitution
cursor.executemany("INSERT INTO users (name, age) VALUES (?, ?)", data)
conn.commit()
````

````python
cursor.execute("SELECT * FROM users")
rows = cursor.fetchall()
for row in rows:
    print(row)
````

````python
import pandas as pd
df = pd.read_sql("SELECT * FROM users", conn)
print(df)
````

## Dropping the table

````python
import sqlite3

# Connect to the database
conn = sqlite3.connect('sample.db')
cursor = conn.cursor()

# Drop the table (this deletes the entire table and all its data)
cursor.execute("DROP TABLE IF EXISTS users")

# Commit the changes
conn.commit()

# Close the connection
cursor.close()
conn.close()
````
