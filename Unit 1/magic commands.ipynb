<h2> Magic commands </h2>

IPython comes with a library of magic commands. These commands are convenient shortcuts to common actions. They all start with % (the percent character). We can get the list of all magic commands with %lsmagic:

```python
%lsmagic
```

```python
%pwd
```

```python
myData = "Python is Amazing!!"
%store myData
# myData= 12365
```

```python
%store -r myData
myData
```

For example, the %%writefile cell magic lets us create a text file easily. This magic command accepts a filename as an argument. All the remaining lines in the cell are directly written to this text file. Here, we create a file test.txt and write Hello world! in it:

```python
%%writefile test.txt
        Hello world!
kjxvjxbc
xmncb xbhgfhytf
    store myData
    gfxcgdgfyhgfydydty
```

#### The File Cursor Is Already at the End
If you opened, read, or modified f earlier in the same notebook session or block, the internal file pointer is at the very end. Calling .read() again returns an empty string "".

```python
with open('test.txt', 'r') as f:
            f.seek(0)
            print(f.read())
```

# File handling

## Opening a File in Python

```python
fobject = open("testfile.txt","w") # creating a data file 
sentence = input("Enter the contents to be written in the file: ") 
fobject.write(sentence) # Writing data to the file 
fobject.close() # Closing a file
```

```python
fobject = open("testfile.txt","r")

with fobject as f:
            print(f.read())

fobject.close()
```

```python
fobject = open("testfile.txt","r")

#looping over the file object to read the file
for st in fobject:
    print(st)
    
fobject.close()
```

## 🔹 Function-level Profiling with `cProfile`

We can measure how much time each function takes. Function-level profiling gives you a top-down view of performance, helping you quickly identify which functions are your primary bottlenecks before you dive into line-by-line optimization.


```python
import cProfile

def slow_function():
    total = 0
    for i in range(1, 1000000):
        total += i
    return total

def fast_function():
    return sum(range(1, 1000000))

print("Profiling slow_function:")
cProfile.run("slow_function()")

print("\nProfiling fast_function:")
cProfile.run("fast_function()")
```
