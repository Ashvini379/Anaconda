# String Indexing

Strings can be indexed (subscripted), with the first character having index 0. There is no separate character type; a character is simply a string of size one:

````python
word = 'Python'
word[0]
````

````python
word[-1]  # last character
````

````python
word[5]  #what will be printing in -0?
````

Note that since -0 is the same as 0, negative indices start from -1.

In addition to indexing, slicing is also supported. While indexing is used to obtain individual characters, slicing allows you to obtain substring:

````python
word[0:5]   # var_name[start:end]
````

````python
word[:9]   # character from the beginning to position 2 (excluded)
````

````python
word[-1:-5]   # characters from position 2 (included) to the end
````

````python
word[-5:-3]   # characters from the second-last (included) to the end
````

````python
s = "hello ... world"


start, end = 1, 100
if start < len(s):
    sliced = s[start:end]
    print(sliced)
else:
    print("Start index out of range")
````

Note how the start is always included, and the end always excluded. This makes sure that s[:i] + s[i:] is always equal to s:

````python
word[:2] + word[2:]
````

Regular expression
By default searches are case sensitive.

```

-[A-Z]	  ----> Finds one uppercase letter.
-[A-Z]{3} ---->	Finds three consecutive uppercase letters.
-[0-9]{5} ---->	Finds five consecutive digits.
-[0-9]+	  ----> Finds one or more digits.
-[^a-z]	  ----> Finds everything except lowercase a to z.
-\s	  ----> Finds one whitespace character (space, tab, and so on).
-\S	  ----> Finds any character except for whitespace.

```

Using RE, try to get the below output for the string CatBatSatFatOr': 
``` 
['Cat', 'Bat', 'Sat', 'Fat', 'Or']
```

````python
import re
str2 = 'CatBatSatFatOr'

res = re.findall('[A-Z][^A-Z]*',str2)
print("Result", str(res))
````
