<h3> Exception Handling</h3>

Exceptions: Exceptions are raised when the program is syntactically correct, but the code resulted in an error.

````python
marks = 10000
 
# perform division with 0
a = marks / 0
print(a)
````

Try and except statements are used to catch and handle exceptions in Python. Statements that can raise exceptions are kept inside the try clause and the statements that handle the exception are written inside except clause.

````python
marks = 10000
a = 0

# perform division with 0
try:
    a = marks / 0

except ZeroDivisionError:
    print("ZeroDivisionError Occurred and Handled")
else:
    print(a)
finally:
    print(a)
````

````python
try:
    print(10 / 2)
    
except ZeroDivisionError:
    print("Division error")
    
finally:
    print("This always runs")
````

````python
try:
    nums = [1, 2, 3]
    print(nums[5])   # invalid index
except IndexError as e:
    print("Type of IndexError is:", e)
finally:
    print("Only 3 elements present in the list")
````

## Using else

````python
num = 12
try:
    num += int("123")
except TypeError:
    print("Invalid number")
else:
    print("Conversion successful:", num)
````

## Catch any Exception

````python
try:
    print(10 / 0)
except Exception as e:
    print("Error occurred:", e)
````

## Using finally

````python
try:
    f = open("sample.txt", "w") # r,w,a
    f.write("Hello sreeja new!")
except Exception as e:
    print("Error:", e)
finally:
    f.close()
    print("File closed")
````
